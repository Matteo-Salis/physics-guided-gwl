# Load and Merge Raw Data

This part consist in loading and merging raw ARPA dataset that are divided by years. Column names are left as in the original datasets

```python
wtd_raw_2000_2004 = pd.read_csv(f"{raw_data_dir}/DatiPiezometriciSoggiacenzaFinoAl2004.csv",
              sep = ";",
              dtype={"objectid":'str',
                     "codice_parametro_arpa":'str',
                     "codice_punto":'str',
                     "gwbregcod":'str'})

wtd_raw_2000_2004["data_ora"] = wtd_raw_2000_2004[["data_misura","ora_misura"]].agg(" ".join, axis=1)

wtd_raw_2000_2004 = wtd_raw_2000_2004.astype({"data_ora":'datetime64[ns]'})

wtd_raw_2000_2004 = wtd_raw_2000_2004[["objectid","codice_parametro_arpa","codice_punto","data_ora", "valore_acquisito", "valore_validato", "gwbregcod"]]
```

```python
wtd_raw_2000_2004
```

```python
wtd_raw_2005_2012 = pd.read_csv(f"{raw_data_dir}/DatiPiezometriciSoggiacenzaDal2005Al2012.csv",
              sep = ";",
              dtype={"objectid":'str',
                     "codice_parametro_arpa":'str',
                     "codice_punto":'str',
                     "gwbregcod":'str'})

wtd_raw_2005_2012["data_ora"] = wtd_raw_2005_2012[["data_misura","ora_misura"]].agg(" ".join, axis=1)

wtd_raw_2005_2012 = wtd_raw_2005_2012.astype({"data_ora":'datetime64[ns]'})

wtd_raw_2005_2012 = wtd_raw_2005_2012[["objectid","codice_parametro_arpa","codice_punto","data_ora", "valore_acquisito", "valore_validato", "gwbregcod"]]
```

```python
wtd_raw_2005_2012
```

```python
wtd_raw_2013_2019 = pd.read_csv(f"{raw_data_dir}/DatiPiezometriciSoggiacenzaDal2013Al2019.csv",
              sep = ";",
              dtype={"objectid":'str',
                     "codice_parametro_arpa":'str',
                     "codice_punto":'str',
                     "gwbregcod":'str'})

wtd_raw_2013_2019["data_ora"] = wtd_raw_2013_2019[["data_misura","ora_misura"]].agg(" ".join, axis=1)

wtd_raw_2013_2019 = wtd_raw_2013_2019.astype({"data_ora":'datetime64[ns]'})

wtd_raw_2013_2019 = wtd_raw_2013_2019[["objectid","codice_parametro_arpa","codice_punto","data_ora", "valore_acquisito", "valore_validato", "gwbregcod"]]
```

```python
wtd_raw_2013_2019
```

```python
wtd_raw_2020_2024 = pd.read_csv(f"{raw_data_dir}/DatiPiezometriciSoggiacenzaDal2020.csv",
              sep = ";",
              dtype={"objectid":'str',
                     "codice_parametro_arpa":'str',
                     "codice_punto":'str',
                     "dataora_misura":'str',
                     "gwbregcod":'str'})

wtd_raw_2020_2024 = wtd_raw_2020_2024.astype({"dataora_misura":'datetime64[ns]'})

wtd_raw_2020_2024 = wtd_raw_2020_2024[["objectid","codice_parametro_arpa","codice_punto","dataora_misura", "valore_acquisito", "valore_validato", "codice_corpo_idrico"]]
```

```python
# Merging
wtd_raw = pd.concat([wtd_raw_2000_2004,
                    wtd_raw_2005_2012,
                    wtd_raw_2013_2019,
                    wtd_raw_2020_2024], ignore_index=True)

wtd_raw = wtd_raw.set_index(['data_ora','codice_punto'])
wtd_raw.index.rename(['data_misura','codice_punto'], inplace = True)
wtd_raw
```

# Spatial Metadata and ROI

```python
underground_sensors_shp = gpd.read_file(f"{geo_data}/underground_wtd_sensor_roi.shp") 
underground_sensors_shp
```

```python
wtd_ds
```

```python
wtd_ds_subset = wtd_ds.loc[pd.IndexSlice[:,underground_sensors_shp["sensor_id"]],:]
wtd_ds_subset = wtd_ds_subset.sort_values(by='date')
wtd_ds_subset
```

```python
wtd_ds_subset = wtd_ds_subset.reset_index().set_index(["date"])
wtd_ds_subset = wtd_ds_subset.groupby([pd.Grouper(freq="D", label = "left"), "sensor_id"]).mean()
wtd_ds_subset = wtd_ds_subset.reset_index()
wtd_ds_subset
```

```python
wtd_ds_subset = wtd_ds_subset.set_index(['date', 'sensor_id'])
```

```python
wtd_ds_subset
```

# Insert missing dates

```python
min_date = str(wtd_ds_subset.index.get_level_values(0).min())
max_date = str(wtd_ds_subset.index.get_level_values(0).max())

filled_date_range = pd.date_range(min_date, max_date, freq="D")
```

```python
wtd_ds_subset.index.get_level_values(1).unique()
```

```python
filled_wtd_ds_subset = wtd_ds_subset.reindex(pd.MultiIndex.from_product((pd.DatetimeIndex(filled_date_range),
                            wtd_ds_subset.index.get_level_values(1).unique()), names=["date", "sensor_id"]),
                     fill_value=np.nan)
```

```python
filled_wtd_ds_subset
```

```python
# Check
print("Old values are the same? ", filled_wtd_ds_subset.loc[wtd_ds_subset.index].equals(wtd_ds_subset))

print("New dates are nan? ", filled_wtd_ds_subset.loc[~filled_wtd_ds_subset.index.isin(wtd_ds_subset.index)].isna().all()) 
```
